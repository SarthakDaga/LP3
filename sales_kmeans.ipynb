```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Load dataset
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")

# 2. Select the four numerical features used for clustering
X = df[["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]]

# 3. Feature scaling
sc = StandardScaler()
X_scaled = sc.fit_transform(X)

# 4. Elbow Method: calculate WCSS for K = 1 to 10
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X_scaled)
    wcss.append(km.inertia_)

plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()

# 5. Final K-Means model (K=4, as in the professor's example)
km = KMeans(n_clusters=4, n_init=10, random_state=42)
df["Cluster"] = km.fit_predict(X_scaled)

# 6. Display results
print("Cluster Counts:")
print(df["Cluster"].value_counts())

print("\nCluster Means:")
print(df.groupby("Cluster")[["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]].mean())
```
